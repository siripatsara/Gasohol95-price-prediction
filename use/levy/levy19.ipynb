```python
import pandas as pd

# อ่าน Excel ทั้งแผ่น
df = pd.read_excel("C:/zzz/oil/scrape/use/levy/P06.xls", header=None)

# แปลงทุกค่าในคอลัมน์ Date เป็น string
df[0] = df[0].astype(str)

# หา index ของปี 2019 และ 2020
start_idx = df[df[0] == '2019'].index[0]
end_idx = df[df[0] == '2020'].index[0]

# เลือกเฉพาะแถวปี 2019
df_2019 = df.iloc[start_idx+1:end_idx, :]

# กรองเฉพาะแถวที่ Date เป็นรูปแบบเดือน/วัน
df_2019 = df_2019[df_2019[0].str.match(r'^[A-Z]{3}-\d{1,2}$')]

# ดึงคอลัมน์ E10 (สมมติอยู่คอลัมน์ 3)
df_e10 = df_2019[[0, 3]].copy()
df_e10.columns = ['Date', 'E10']

# แปลง Date เป็น datetime
df_e10['Date'] = pd.to_datetime(df_e10['Date'] + '-2019', errors='coerce', format='%b-%d-%Y')

# ลบแถวที่แปลง Date ไม่ได้
df_e10 = df_e10.dropna(subset=['Date'])

# แปลง E10 เป็น numeric และเติม NaN ด้วย interpolate
df_e10['E10'] = pd.to_numeric(df_e10['E10'], errors='coerce')
df_e10['E10'] = df_e10['E10'].interpolate(method='linear')

# ตรวจสอบจำนวนแถวและ 5 แถวแรก
print(df_e10.shape)
print(df_e10.head())
print(df_e10.tail())
```

```python
# --- สร้าง DataFrame ครบทุกวันปี 2019 ---
full_dates = pd.date_range(start='2019-01-01', end='2019-12-31', freq='D')
df_full = pd.DataFrame({'Date': full_dates})

# --- merge ข้อมูลสัปดาห์ลงทุกวัน ---
df_full = df_full.merge(df_e10, on='Date', how='left')

# --- เติมค่า E10 ด้วย forward fill ---
df_full['E10'] = df_full['E10'].fillna(method='ffill')

# --- ตรวจสอบ 10 แถวแรกและ 10 แถวสุดท้าย ---
print(df_full.head(10))
print(df_full.tail(10))
```

```python
# --- save CSV ---
df_full.to_csv("C:/zzz/oil/scrape/use/levy/data/levy2019.csv", index=False, encoding="utf-8-sig")
```
