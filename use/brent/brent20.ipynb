```python
import pandas as pd
```

```python
df = pd.read_csv("C:/zzz/oil/scrape/use/brent/2020/DCOILBRENTEU20.csv")
df
```

```python
df = df[["observation_date","DCOILBRENTEU"]].rename(columns={"observation_date":"Date","DCOILBRENTEU":"Price_Brent"})
df
```

```python
import pandas as pd

# สมมติ df มีคอลัมน์ 'date' และ 'Price_Brent'
df['Date'] = pd.to_datetime(df['Date'])

# สร้างช่วงวันที่ครบทุกวัน
full_dates = pd.date_range(start='2020-01-01', end='2020-12-31', freq='D')

# ตั้ง DataFrame ใหม่โดยใช้ full_dates
df_full = pd.DataFrame({'Date': full_dates})

# Merge กับข้อมูลเดิม (left join) เพื่อให้มีทุกวัน
df_full = df_full.merge(df, on='Date', how='left')

# เติมค่า Price_Brent ที่เป็น NaN ด้วย linear interpolation
df_full['Price_Brent'] = df_full['Price_Brent'].interpolate(method='linear')

# ตรวจสอบ
df_full
```

```python
df_full.isna().sum()
```

```python
nan_rows = df_full[df_full["Price_Brent"].isna()]

nan_rows
```

```python
import pandas as pd

# หา index ของแถวที่เป็น NaN
nan_index = df_full[df_full["Price_Brent"].isna()].index

# สร้าง list ของ index ที่ต้องการ (ก่อนหน้า + แถว NaN + หลัง)
rows_to_print = []
for i in nan_index:
    if i-1 >= 0:
        rows_to_print.append(i-1)  # แถวก่อนหน้า
    rows_to_print.append(i)        # แถวที่เป็น NaN
    if i+1 < len(df_full):
        rows_to_print.append(i+1)  # แถวหลัง

# เอาเฉพาะ index ไม่ซ้ำ
rows_to_print = sorted(set(rows_to_print))

# แสดงผล
df_full.loc[rows_to_print]
```

```python
# 67.77

# เติมค่าแถวแรกด้วยค่าก่อนหน้า (67.77)
df_full.loc[0, 'Price_Brent'] = 67.77

# เติมค่า NaN ที่เหลือด้วย linear interpolation
df_full['Price_Brent'] = df_full['Price_Brent'].interpolate(method='linear')

# ตรวจสอบ 5 แถวแรก
df_full
```

```python
# ตั้งชื่อไฟล์ brent 2020 เก็บไฟล์บน C:\zzz\oil\scrape\use\brent\data

import os

# สร้างโฟลเดอร์ถ้ายังไม่มี
output_dir = r"C:\zzz\oil\scrape\use\brent\data"
os.makedirs(output_dir, exist_ok=True)

# กำหนด path ของไฟล์
output_file = os.path.join(output_dir, "brent2020.csv")

# บันทึกเป็น CSV
df_full.to_csv(output_file, index=False)

print(f"Exported CSV to: {output_file}")
```
